**ENCAPSULATION IN PYTHON**

```python
class bankAccount:
    def __init__(self,name,__balance):
        self.name = name
        self.__balance = __balance

    #getter
    def display(self):
        return self.__balance

    #setter
    def deposit(self,amount_depo):
        self.__balance += amount_depo
        print("Amount Diposited. Updated Balance = ",self.__balance)

    def withdrawn(self,amount_with):
        if amount_with > self.__balance:
            print("Insufficient Balance")
            return
        self.__balance -= amount_with
        print("Amount Withdrawn. Updated Balance = ",self.__balance)
    

a1 = bankAccount('Rituparna',1000)
a1.deposit(5000)
a1. withdrawn(2500)
```

```python
class InstaAccount:
    def __init__(self,username,__password):
        self.username = username
        self.__password = __password

    def updatePassword(self,newPassword):
        self.__password = newPassword

u1 = InstaAccount('Rituparna',56789)
u1.updatePassword(37914)
```

```python
class Student:
    def __init__(self,name,__marks):
        self.name = name
        self.__marks = __marks

    @property
    def marks(self):
        return self.__marks
    
    @marks.setter
    def marks(self,newmarks):
            self.__marks = newmarks

s1 = Student("Subha",50)
print(s1.marks)
s1.marks = 98
print(s1.marks)
```

**Abstraction In Python**
